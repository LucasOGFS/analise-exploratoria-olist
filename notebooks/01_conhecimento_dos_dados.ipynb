```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

sns.set_theme(style="whitegrid")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
caminho_dados = Path("../data/raw")

arquivos = list(caminho_dados.glob("*.csv"))

print(f"Quantidade de arquivos encontrados: {len(arquivos)}")

for arquivo in arquivos:
    print(arquivo.name)
```

```python
dados = {}

for arquivo in arquivos:
    nome = arquivo.stem.replace("_dataset", "")
    dados[nome] = pd.read_csv(arquivo, encoding="utf-8", low_memory=False)

print("Todos os arquivos foram carregados!")
```

```python
for nome, df in dados.items():
    print(f"\nTabela: {nome}")
    print(f"Linhas: {df.shape[0]}")
    print(f"Colunas: {df.shape[1]}")
```

```python
for nome, df in dados.items():
    print(f"\nTabela: {nome}")
    ausentes = df.isnull().sum()
    ausentes = ausentes[ausentes > 0]

    if ausentes.empty:
        print("Nenhum valor ausente")
    else:
        print(ausentes)
```

```python
for nome, df in dados.items():
    duplicados = df.duplicated().sum()
    print(f"{nome}: {duplicados} registros duplicados")
```
